```python
from pathlib import Path

from sentinela.bootstrap import (
    prepare_sentinela_environment,
)

from sentinela.prospective import (
    run_prospective_week,
)
```

```python
env = prepare_sentinela_environment(
    influd_path=Path(
        "../data/raw/INFLUD26-19-10-2026.csv"
    ),
    year=2026,
)

results = run_prospective_week(
    prospective_weekly=env.weekly,
    epi_year=env.epi_year,
    current_week=env.latest_ready_week,
)
```

```python
print(
    "Cutoff:",
    env.latest_ready_week,
)

print(
    "Eventos:",
    len(results),
)

print(
    "UFs:",
    results["region"].nunique(),
)

print(
    "Duplicatas:",
    results[
        [
            "region",
            "change_week",
            "direction",
            "llr",
        ]
    ]
    .duplicated()
    .sum(),
)
```
